### Python list

- lists are mutable 
- tạo một list mới: my_list = list()
- `append()` – Thêm 1 phần tử vào cuối list.
  - Cú pháp: `list.append(element)`
  - Ví dụ: `a = [1, 2]; a.append(3)  # [1, 2, 3]`
  - Độ phức tạp: **O(1)** (trung bình)

- `extend()` – Thêm các phần tử của một **iterable** vào cuối list.
  - Cú pháp: `list.extend(iterable)`
  - Ví dụ: `a = [1, 2]; a.extend([3, 4])  # [1, 2, 3, 4]`
  - Độ phức tạp: **O(k)** (`k` là số phần tử được thêm)

- `sort()` / `sorted()` – Sắp xếp list.
  - Cú pháp: `list.sort()` hoặc `sorted(list)`
  - Ví dụ: `a = [3, 1, 2]; a.sort()  # [1, 2, 3]`
  - Độ phức tạp: **O(n log n)**

- `reverse()` – Đảo ngược thứ tự các phần tử.
  - Cú pháp: `list.reverse()`
  - Ví dụ: `a = [1, 2, 3]; a.reverse()  # [3, 2, 1]`
  - Độ phức tạp: **O(n)**

- `insert()` – Chèn phần tử vào vị trí chỉ định.
  - Cú pháp: `list.insert(index, value)`
  - Ví dụ: `a = [1, 3]; a.insert(1, 2)  # [1, 2, 3]`
  - Độ phức tạp: **O(n)**

- `del` / `remove()` – Xóa phần tử.
  - Cú pháp: `del list[index]` hoặc `list.remove(value)`
  - Ví dụ: `a = [1, 2, 3]; a.remove(2)  # [1, 3]`
  - Độ phức tạp: **O(n)**

- `index()` – Trả về vị trí đầu tiên của phần tử.
  - Cú pháp: `list.index(value)`
  - Ví dụ: `a = [5, 2, 8]; a.index(2)  # 1`
  - Độ phức tạp: **O(n)**

- `pop()` – Xóa và trả về phần tử.
  - Cú pháp: `list.pop(index)` hoặc `list.pop()`
  - Ví dụ: `a = [1, 2, 3]; a.pop()  # trả về 3, a = [1, 2]`
  - Độ phức tạp:
    - `pop()` cuối list: **O(1)**
    - `pop(index)`: **O(n)**

### pickle in python

- overview:
    - dùng để lưu trữ(serialize) và truyền tải(deserialize) data một cấu trúc của đối tượng
    - modul pickle
- ưu điểm:
    - lưu trực tiếp các obj python( list, dict, ...) mà không cần chuyển sang chuỗi
    - dễ sử dụng dùng dump() để lưu và load() để đọc
- nhược điểm:
    - không an toàn: không nên pick.load() từ file không tin cậy nó có thể chứa mã độc
    


````python
import pickle 

my_dict = {
    "name" : "hoang",
    "age" : 20,
    "school" : "uet"
}
````

````python
# ghi mot obj vao file
with open("my_pickle.pickle", "wb") as file:
    pickle.dump(my_dict, file)
````

````python
# doc mot obj tu file
with open("my_pickle.pickle", "rb") as file:
    text = pickle.load(file)
print(text)
print(type(text))
````

````python
# chuyển obj thành bytes
my_binary = pickle.dumps(my_dict)
print(my_binary)
print(type(my_binary))
````

````python
# chuyển bytes thành obj
old_dict = pickle.loads(my_binary)
print(old_dict)
print(type(old_dict))
````

### Json( JavaScript object notation)
- JSON (JavaScript Object Notation) là một định dạng trao đổi dữ liệu mới và ngày càng phổ biến. JSON không phụ thuộc vào ngôn ngữ lập trình và được hỗ trợ bởi nhiều hệ thống khác nhau. Ví dụ, MongoDB lưu trữ dữ liệu dưới dạng tài liệu JSON (JSON document).
- JSON có cú pháp rất giống với dictionary và list trong Python, vì vậy việc chuyển đổi từ object Python sang JSON và ngược lại rất đơn giản.
- Module sử dụng: json

````python
import json

my_dict = {
    "name" : "Hoang",
    "age" : 20, 
    "hobbies": ["bong da", "game"]
}
print(my_dict)
print(type(my_dict))
````

````python
# chuyen obj dict python sang json string

json_string = json.dumps(my_dict)
print(json_string)
print(type(json_string))
````

````python
# chuyen tu json string sang dict

new_dict = json.loads(json_string)
print(new_dict)
print(type(new_dict))
````

````python
# kiểm tra my_dict và new_dict có giống nhau không
assert new_dict == my_dict
print("OK")
````

````python
# ghi python object vào một file .json
with open("save_json.json", "w", encoding = "utf-8") as file:
    json.dump(my_dict, file, indent= 4)
````

````python
# đọc file json về python dict
with open("save_json.json") as file:
    text = json.load(file)
print(text)
print(type(text))
````

````python
#filter(func, iterable): dòng để lọc các phần tử trong một iterable theo một điều kiện
lst = [1, 2, 3, 4, 5, 6]
new_list = (filter(lambda x: x%2 == 0, lst))
print(new_list)
````

````python
#reduce(func, iterable): dùng để gộp tất cả các phần tử của một iterable thanh một giá trị duy nhất
lst = [1, 2, 3, 4, 5, 6]
from functools import reduce
new_list = reduce(lambda x, y: x ** y, lst)
print(new_list)
````

##### 1.1. Class Method là gì?

**Class Method** là phương thức thuộc về **class**, không thuộc về từng object.

Nó được khai báo bằng decorator:

```python
@classmethod
```

Tham số đầu tiên của class method luôn là:

```python
cls
```

Trong đó:

- `cls` là tham chiếu đến chính class hiện tại.
- Theo quy ước người ta đặt tên là `cls`, tuy nhiên có thể đặt tên khác (không nên).

Ví dụ:

```python
class Car:

    @classmethod
    def hello(cls):
        print(cls)
```

Gọi:

```python
Car.hello()
```

Kết quả:

```text
<class '__main__.Car'>
```

Điều này chứng tỏ `cls` chính là class `Car`.

##### 1.2. Cú pháp

```python

        ...
```

Ví dụ:

```python

```

**Trường hợp 2: Alternative Constructor**

Alternative Constructor là cách tạo object bằng nhiều kiểu dữ liệu khác nhau.

Ví dụ:

```python
class Car:

    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    @classmethod
    def from_string(cls, text):
        brand, model = text.split(",")
        return cls(brand, model)
```

Sử dụng:

```python
car = Car.from_string("BMW,i320")
```

Python sẽ thực hiện:

```python
Car("BMW", "i320")
```

Ví dụ khác:

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    @classmethod
    def from_list(cls, data):
        return cls(data[0], data[1])
```

```python
student = Student.from_list(["Hoàng", 20])
```

Sử dụng:

#### 2. Static Method

##### 2.1. Static Method là gì?

**Static Method** là phương thức thuộc về **class**, nhưng **không làm việc trực tiếp với class hay object**.

Nó được khai báo bằng decorator:

```python
@staticmethod
```

Static Method **không có** tham số `self` và cũng **không có** tham số `cls`.

Ví dụ:

```python
class Math:

    @staticmethod
    def add(a, b):
        return a + b
```

Gọi:

```python
print(Math.add(3, 5))
```

Kết quả:

```text
8
```

##### 2.2. Cú pháp

```python
class ClassName:

    @staticmethod
    def method_name(arg1, arg2):
        ...
```

Ví dụ:

```python
class Calculator:

    @staticmethod
    def multiply(a, b):
        return a * b
```

Sử dụng:

```python
result = Calculator.multiply(5, 4)

print(result)
```

Kết quả:

```text
20
```

##### 2.3. Static Method có gì đặc biệt?

Static Method:

- Không có `self`.
- Không có `cls`.
- Không truy cập trực tiếp được Instance Variable.
- Không truy cập trực tiếp được Class Variable.
- Chỉ hoạt động với các tham số được truyền vào.

Ví dụ:

```python
class Math:

    @staticmethod
    def square(x):
        return x * x
```

```python
print(Math.square(6))
```

Kết quả:

```text
36
```

##### 2.4. Có thể gọi Static Method bằng Object không?

Có.

Ví dụ:

```python
class Math:

    @staticmethod
    def add(a, b):
        return a + b
```

```python
math = Math()

print(Math.add(2, 3))
print(math.add(2, 3))
```

Kết quả:

```text
5
5
```

Tuy nhiên, nên gọi bằng tên class:

```python
Math.add(2, 3)
```

vì Static Method thuộc về class.

##### 2.5. Static Method có thể truy cập Class Variable không?

Không trực tiếp.

Ví dụ:

```python
class Car:

    tax = 1.1

    @staticmethod
    def show():
        print(tax)
```

Kết quả:

```text
NameError
```

Muốn truy cập phải ghi rõ tên class.

Ví dụ:

```python
class Car:

    tax = 1.1

    @staticmethod
    def show():
        print(Car.tax)
```

Kết quả:

```text
1.1
```

##### 2.6. Static Method có thể truy cập Instance Variable không?

Không.

Ví dụ:

```python
class Car:

    def __init__(self):
        self.brand = "BMW"

    @staticmethod
    def show():
        print(self.brand)
```

Kết quả:

```text
NameError
```

Lý do:

Static Method không có `self`.

##### 2.7. Khi nào dùng Static Method?

Static Method thường dùng cho các **hàm tiện ích (Utility Function)**.

Đó là những hàm:

- Không cần object.
- Không cần class.
- Chỉ thực hiện tính toán hoặc xử lý dữ liệu.

Ví dụ:

```python
class Math:

    @staticmethod
    def max(a, b):
        return a if a > b else b

    @staticmethod
    def min(a, b):
        return a if a < b else b
```

Sử dụng:

```python
print(Math.max(7, 9))
print(Math.min(7, 9))
```

Kết quả:

```text
9
7
```

##### 2.8. Ví dụ thực tế

Ví dụ kiểm tra số chẵn.

```python
class Number:

    @staticmethod
    def is_even(n):
        return n % 2 == 0
```

```python
print(Number.is_even(10))
print(Number.is_even(15))
```

Kết quả:

```text
True
False
```

Ví dụ kiểm tra email.

```python
class Validator:

    @staticmethod
    def is_gmail(email):
        return email.endswith("@gmail.com")
```

```python
print(Validator.is_gmail("abc@gmail.com"))
```

Kết quả:

```text
True
```

Ví dụ tính diện tích hình tròn.

```python

    PI = 3.14159

    @staticmethod
    def circle_area(radius):
        return Geometry.PI * radius ** 2
```

```python

```

Kết quả:

```text

```

##### 2.9. Static Method hoạt động như thế nào?

Ví dụ:

```python

    @staticmethod
    def add(a, b):
        return a + b
```

Khi gọi:

```python

```

Python thực hiện:

```python

```

Không truyền thêm `self` hay `cls`.

Có thể hình dung:

```text

↓

```

##### 2.10. Phân biệt Instance Method, Class Method và Static Method

| Đặc điểm | Instance Method | Class Method | Static Method |
|----------|-----------------|--------------|---------------|
| Decorator | Không có | `@classmethod` | `@staticmethod` |
| Tham số đầu tiên | `self` | `cls` | Không có |
| Thuộc về | Object | Class | Class |
| Truy cập Instance Variable | ✅ | ❌ | ❌ |
| Truy cập Class Variable | ✅ | ✅ | Chỉ khi ghi rõ tên class |
| Có thể tạo Object | ✅ | ✅ | ❌ |
| Có thể gọi bằng Object | ✅ | ✅ | ✅ |
| Có thể gọi bằng Class | ❌ (không nên) | ✅ | ✅ |

##### 2.11. Khi nào nên dùng?

Sử dụng **Instance Method** khi phương thức làm việc với dữ liệu của từng object.

Sử dụng **Class Method** khi phương thức làm việc với dữ liệu chung của class hoặc tạo object theo nhiều cách khác nhau.

Sử dụng **Static Method** khi phương thức chỉ là một hàm tiện ích, không cần truy cập object hay class.

##### 2.12. Ví dụ tổng hợp

```python
class Car:

    tax = 1.1

    def __init__(self, brand):
        self.brand = brand

    def show_brand(self):
        print(self.brand)

    @classmethod
    def show_tax(cls):
        print(cls.tax)

    @staticmethod
    def km_to_mile(km):
        return km * 0.621371
```

Sử dụng:

```python
car = Car("BMW")

car.show_brand()

Car.show_tax()

print(Car.km_to_mile(100))
```

Kết quả:

```text
BMW
1.1
62.1371
```

##### 2.13. Mẹo ghi nhớ

- `self` → làm việc với **object**.
- `cls` → làm việc với **class**.
- Không có `self`, không có `cls` → chỉ là **một hàm tiện ích**, đó chính là **Static Method**.


# Buổi 5: Các Method Đặc Biệt (Magic Methods / Dunder Methods)

## Mục tiêu

Sau buổi học này bạn sẽ:

- Hiểu Magic Method là gì.
- Biết Python tự động gọi Magic Method khi nào.
- Phân biệt `__repr__()` và `__str__()`.
- Biết cách nạp chồng toán tử bằng `__add__()`.
- Hiểu cách hoạt động của `__len__()`.
- Hiểu `__del__()` và cơ chế hủy object.
- So sánh các Magic Method với Java.

---

# 1. Magic Method là gì?

Magic Method (hay còn gọi là **Dunder Method**) là những phương thức có dạng:

```python
__ten_phuong_thuc__
```

Ví dụ:

```python

__str__()
__repr__()
__add__()
__len__()
```

Python sẽ **tự động gọi** các method này khi object tham gia vào một hành động đặc biệt.

Ví dụ:

```python
print(obj)
```

Python thực chất gọi

```python
obj.__str__()
```

---

```python
len(obj)
```

Python thực chất gọi

```python
obj.__len__()
```

---

```python
a + b
```

Python thực chất gọi

```python
a.__add__(b)
```

---

## Tại sao cần Magic Method?

Nếu không có Magic Method thì object chỉ là một vùng nhớ đơn thuần.

    pass

# 2. __repr__(self)

- Hỗ trợ debug.
- Hiển thị thông tin chi tiết.

def __repr__(self):

---

# 9. Tóm tắt

|---------|--------------|----------|

| `__len__()` | `len(obj)` | Trả về độ dài object |

# Kiến thức cần nhớ

- Magic Method được Python tự động gọi.
- Không nên tự gọi trực tiếp `__str__()` hay `__add__()`, hãy dùng `print(obj)`, `a + b`, `len(obj)`... để Python tự xử lý.
- `__repr__()` dành cho lập trình viên, `__str__()` dành cho người dùng.
- `__str__()` được ưu tiên khi `print(obj)`.
- `__add__()` giúp object hỗ trợ toán tử `+`.
- `__len__()` cho phép dùng `len()` với object.
- `__del__()` không đảm bảo thời điểm thực thi, nên hạn chế sử dụng trong các chương trình thực tế.

# Buổi 6: Property, Getter, Setter, Deleter

## Mục tiêu

Sau buổi học này bạn sẽ:

- Hiểu Property là gì.
- Biết cách tạo Getter.
- Biết cách tạo Setter.
- Biết cách tạo Deleter.
- Hiểu tại sao Python khuyến khích dùng Property thay vì Getter/Setter truyền thống.
- So sánh Property trong Python với Java.

---

# 1. Property là gì?

## Khái niệm

Property là một cơ chế trong Python cho phép **truy cập một phương thức như thể nó là một thuộc tính**.

Nói đơn giản:

Thay vì gọi

```python
student.get_name()
```

ta có thể viết

```python
student.name
```

nhưng phía sau Python vẫn thực hiện một phương thức.

Property giúp:

- Đọc dữ liệu
- Kiểm tra dữ liệu
- Thay đổi dữ liệu
- Xóa dữ liệu

mà người dùng vẫn thao tác giống như với một thuộc tính bình thường.

---

## Tại sao cần Property?

Giả sử có class

```python
class Student:

    def __init__(self, name):
        self.name = name
```

Có thể sửa trực tiếp

```python
student.name = "Hoang"
```

Nếu muốn kiểm tra dữ liệu

Ví dụ

- Không cho tên rỗng
- Không cho tuổi âm
- Không cho điểm > 10

thì cách trên không làm được.

Property giúp giải quyết vấn đề này.

---

# 2. Getter

## Khái niệm

Getter là phương thức dùng để lấy giá trị của thuộc tính.

Trong Python sử dụng decorator

```python
@property
```

---

## Cú pháp

```python
@property
def name(self):
    return ...
```

---

## Ví dụ

```python
class Student:

    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        return self._name

student = Student("Hoang")

print(student.name)
```

Kết quả

```
Hoang
```

---

## Python hoạt động như thế nào?

Khi viết

```python
student.name
```

Python thực chất gọi

```python
student.name()
```

nhưng tự động.

Lập trình viên không cần thêm dấu ngoặc.

---

## Lợi ích

- Đọc dữ liệu dễ dàng.
- Có thể xử lý trước khi trả về.
- Có thể tính toán giá trị.

Ví dụ

```python
@property
def full_name(self):
    return self.first_name + " " + self.last_name
```

---

# 3. Setter

## Khái niệm

Setter là phương thức dùng để thay đổi giá trị của Property.

Sử dụng

```python
@ten_property.setter
```

---

## Cú pháp

```python
@name.setter
def name(self, value):
    ...
```

---

## Ví dụ

```python
class Student:

    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        return self._name

    @name.setter
    def name(self, value):
        self._name = value

student = Student("Hoang")

student.name = "An"

print(student.name)
```

Kết quả

```
An
```

---

## Validation

Setter thường dùng để kiểm tra dữ liệu.

Ví dụ

```python
class Student:

    def __init__(self, age):
        self._age = age

    @property
    def age(self):
        return self._age

    @age.setter
    def age(self, value):

        if value < 0:
            raise ValueError("Age phải >= 0")

        self._age = value
```

---

Ví dụ

```python
student = Student(20)

student.age = -5
```

Kết quả

```
ValueError
```

---

## Một ví dụ khác

```python
class Product:

    def __init__(self, price):
        self._price = price

    @property
    def price(self):
        return self._price

    @price.setter
    def price(self, value):

        if value <= 0:
            raise ValueError("Price phải lớn hơn 0")

        self._price = value
```

---

# 4. Deleter

## Khái niệm

Deleter là phương thức được gọi khi sử dụng

```python
del object.property
```

Sử dụng

```python
@property_name.deleter
```

---

## Cú pháp

```python
@name.deleter
def name(self):
    ...
```

---

## Ví dụ

```python
class Student:

    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        return self._name

    @name.deleter
    def name(self):
        del self._name
```

---

Sử dụng

```python
student = Student("Hoang")

del student.name
```

---

Sau đó

```python
print(student.name)
```

Sẽ báo lỗi

```
AttributeError
```

---

## Có thể dùng để làm gì?

- Xóa dữ liệu.
- Ghi log.
- Giải phóng tài nguyên.
- Thông báo object đã thay đổi.

Ví dụ

```python
@name.deleter
def name(self):

    print("Đã xóa tên")

    del self._name
```

---

# 5. Getter, Setter truyền thống

Python vẫn hỗ trợ.

Ví dụ

```python
class Student:

    def __init__(self, name):
        self._name = name

    def get_name(self):
        return self._name

    def set_name(self, value):
        self._name = value
```

Sử dụng

```python
student.get_name()

student.set_name("An")
```

---

## Tại sao ít dùng?

Python khuyến khích

```python
student.name
```

thay vì

```python
student.get_name()
```

Code ngắn hơn.

Đẹp hơn.

Dễ đọc hơn.

Chỉ có Getter.

Không có Setter.

Được.

Có Getter.

Có Setter.

# 12. Khi nào nên dùng Property?

Nên dùng khi

- Kiểm tra dữ liệu.
- Muốn thuộc tính chỉ đọc.
- Muốn tính toán trước khi trả về.
- Muốn giữ giao diện đơn giản.
- Muốn thay đổi cách lưu dữ liệu mà không ảnh hưởng đến code bên ngoài.

---

# 13. Khi nào không cần Property?

Không cần khi

- Thuộc tính đơn giản.
- Không cần kiểm tra dữ liệu.
- Chỉ là class dùng nội bộ.

Ví dụ

```python
class Point:

    def __init__(self,x,y):
        self.x=x
        self.y=y
```

Không cần tạo Property.

|------------|-----------|
| `@property` | Getter |
| `@property_name.setter` | Setter |
| `@property_name.deleter` | Deleter |

# So sánh Java

| Java | Python |
|-------|---------|
| `getName()` | `@property` |

# Kiến thức cần nhớ

- Property là cách Python biến một phương thức thành thuộc tính.
- `@property` dùng để tạo Getter.
- `@property_name.setter` dùng để tạo Setter.
- `@property_name.deleter` dùng để tạo Deleter.
- Property giúp kiểm tra dữ liệu mà vẫn giữ cú pháp truy cập đơn giản.
- Dữ liệu thật thường được lưu trong biến có tiền tố `_`, ví dụ `_name`, để tránh đệ quy vô hạn.
- Python ưu tiên Property hơn Getter/Setter truyền thống vì cú pháp ngắn gọn, tự nhiên và dễ đọc.
