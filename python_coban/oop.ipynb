#### bài 1
Thiết kế Xe hơi: Viết một class Car với hàm __init__ nhận vào 2 tham số là brand (hãng xe) và model. Bên trong __init__, hãy gán chúng cho các thuộc tính self.brand và self.model. Sau đó, tạo ra 2 đối tượng xe hơi khác nhau và in ra thông tin của chúng.

```python
class car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
    def __str__(self):
        return f"xe cua hang {self.brand}, xe loai {self.model}"
        
car_1 = car("vinfast", "xe dien")
car_2 = car("honda", "xe may")

print(car_1)
print(car_2)
```
