## Cây quyết định Decision tree
### 1, Ví dụ
- sắp đến kì thi, một cậu học sinh đưa ra quyết định nên chơi hay học. 

-> việc quan sát và đưa ra quyết định của con người thường được bắt đầu từ các câu hỏi. ML cũng có một mô hình đưa ra quyết định dựa trên câu hỏi. Mô hình này được gọi là cây quyết định
- xét ví dụ trên trong không gian 2 chiều với 2 class là lục và đỏ. Nhiệm vụ là đi tìm ranh giới đơn giản giữa 2 class này nói cách khác đây là bài toán phân loại( classification)


- Trong decision tree:
    - các ô màu xám, lục, đỏ trên hình 2 còn được gọi là các node. Các node thể hiện đầu ra( đỏ và xanh) còn được gọi là node lá hay leaf node hoặc terminal node.
    - các node thể hiện câu hỏi là non-leaf node. None-leaf node trên cùng được gọi là node gốc( root node)
    - các child node có thể là leaf node hoặc non-leaf node. Các child node có cùng bố mẹ còn được gọi là sibling node
    - lưu ý: các non leaf node chỉ có 2 child node được gọi là một binary decison. VD: ta có thể xác định tuổi của một người dựa trên nhiều câu hỏi đúng sai -> đây là thuật toán tìm kiếm nhị phân
    
- Decision tree là một mô hình supervised learning, có thể áp dụng vào cả 2 thuật toán classification và regression. Việc xây dựng thuật toán dựa trên tập dữ liệu huấn luyện cho trước là việc ta đi xác định các câu hỏi và thứ tự của chúng.

#### 2.1, Ý tưởng
- chúng ta cần xác định thứ tự của các thuộc tính cần được xem xét ở mỗi bước. Khi mà bài toán có nhiều thuộc tính và mỗi thuộc tính có nhiều giá trị khác nhau thì việc tìm ra nghiệm tối ưu cho bài toán là không khả thi. Nên là ở mỗi bước chúng ta sẽ chọn ra một thuộc tính tốt nhất dựa trên một tiêu chuẩn yêu cầu nào đó. với mỗi thuộc tính được chọn ta chia dữ liệu vào childnode tương ứng
- thế nào là một phép phân chia? Bằng trực giác, một phép phân là tốt nhất nếu dữ liệu trong mỗi child node hoàn toàn thuộc vào một class- khi đó child node này có thể coi là leaf node. Từ nhận xét này ta cần có một hàm số đo độ tinh khiết(purity ), hoặc độ vẩn đục(impurity ) của một phép phân chia. Hàm số này sẽ cho giá trị thấp nhất nếu các child node cùng nằm trong một class.

#### 2.2, Hàm số entropy 
- entropy là đại lượng đo lường độ hỗn loạn, độ bất định hoặc độ vẩn đục của một tập dữ liệu
- cho một phân phối xác xuất biến rời rạc x có thể nhận n giá trị khác nhau $ x_1, x_2, ..., x_n$. giả sử xác suất để x nhận giá trị này là $p_i = p(x = xi)$ với $0 \le p_i \le 1, \sum_{i = 1}^{n} p_i = 1$. Ký hiệu phân hối này là $ \mathbf{p} = {p_1, p_2, ..., p_n}$ :
$$
H(\mathbf{p}) = -\sum_{i = 1}^{n} p_i \log(p_i)
$$
